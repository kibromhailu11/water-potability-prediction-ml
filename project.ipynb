```python
# ============================================================
# ML - CLASSIFICATION IMPLEMENTATION
# Pulsar Stars Dataset
# K-Nearest Neighbors (K-NN)
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    confusion_matrix,
    roc_curve,
    roc_auc_score,
    auc,
    classification_report
)

%matplotlib inline


# ============================================================
# 2. LOAD DATASET
# ============================================================

data = 'data/pulsar_stars.csv'
df = pd.read_csv(data)


# ============================================================
# 3. BASIC DATA EXPLORATION
# ============================================================

# Check shape of dataset
print("Dataset shape:")
print(df.shape)

# Display first 5 rows
print("\nFirst 5 rows:")
display(df.head())

# Display column names
print("\nOriginal column names:")
print(df.columns)


# ============================================================
# 4. CLEAN COLUMN NAMES
# ============================================================

# Remove leading and trailing spaces
df.columns = df.columns.str.strip()

print("\nColumn names after removing spaces:")
print(df.columns)


# ============================================================
# 5. RENAME COLUMN NAMES
# ============================================================

df.columns = [
    'IP Mean',
    'IP Sd',
    'IP Kurtosis',
    'IP Skewness',
    'DM-SNR Mean',
    'DM-SNR Sd',
    'DM-SNR Kurtosis',
    'DM-SNR Skewness',
    'target_class'
]

print("\nRenamed column names:")
print(df.columns)


# ============================================================
# 6. CHECK TARGET VARIABLE DISTRIBUTION
# ============================================================

print("\nTarget class counts:")
print(df['target_class'].value_counts())

print("\nTarget class percentages:")
print(df['target_class'].value_counts() / float(len(df)))


# Optional visualization of target distribution

plt.figure(figsize=(6, 4))

sns.countplot(
    x='target_class',
    data=df
)

plt.title('Target Class Distribution')
plt.xlabel('Target Class')
plt.ylabel('Count')

plt.show()


# ============================================================
# 7. DATASET INFORMATION
# ============================================================

print("\nDataset information:")
df.info()


# ============================================================
# 8. CHECK FOR MISSING VALUES
# ============================================================

print("\nMissing values:")
print(df.isnull().sum())


# ============================================================
# 9. SUMMARY STATISTICS
# ============================================================

print("\nSummary statistics:")
display(round(df.describe(), 2))


# ============================================================
# 10. BOX PLOTS - CHECK FOR OUTLIERS
# ============================================================

plt.figure(figsize=(24, 20))

plt.subplot(4, 2, 1)
fig = df.boxplot(column='IP Mean')
fig.set_title('')
fig.set_ylabel('IP Mean')

plt.subplot(4, 2, 2)
fig = df.boxplot(column='IP Sd')
fig.set_title('')
fig.set_ylabel('IP Sd')

plt.subplot(4, 2, 3)
fig = df.boxplot(column='IP Kurtosis')
fig.set_title('')
fig.set_ylabel('IP Kurtosis')

plt.subplot(4, 2, 4)
fig = df.boxplot(column='IP Skewness')
fig.set_title('')
fig.set_ylabel('IP Skewness')

plt.subplot(4, 2, 5)
fig = df.boxplot(column='DM-SNR Mean')
fig.set_title('')
fig.set_ylabel('DM-SNR Mean')

plt.subplot(4, 2, 6)
fig = df.boxplot(column='DM-SNR Sd')
fig.set_title('')
fig.set_ylabel('DM-SNR Sd')

plt.subplot(4, 2, 7)
fig = df.boxplot(column='DM-SNR Kurtosis')
fig.set_title('')
fig.set_ylabel('DM-SNR Kurtosis')

plt.subplot(4, 2, 8)
fig = df.boxplot(column='DM-SNR Skewness')
fig.set_title('')
fig.set_ylabel('DM-SNR Skewness')

plt.tight_layout()
plt.show()


# ============================================================
# 11. SEPARATE FEATURES (X) AND TARGET (y)
# ============================================================

X = df.drop(['target_class'], axis=1)

y = df['target_class']

print("\nFeatures (X):")
display(X.head())

print("\nTarget (y):")
display(y.head())


# ============================================================
# 12. TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=0,
    stratify=y
)

print("\nTraining set shape:")
print(X_train.shape)

print("\nTesting set shape:")
print(X_test.shape)


# ============================================================
# 13. FEATURE SCALING
# ============================================================

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)

print("\nFeature scaling completed.")


# ============================================================
# 14. TRAIN INITIAL K-NN CLASSIFIER
# ============================================================

knn = KNeighborsClassifier(
    n_neighbors=3
)

knn.fit(
    X_train,
    y_train
)

print("\nInitial K-NN model:")
print(knn)


# ============================================================
# 15. MAKE PREDICTIONS
# ============================================================

y_pred_knn = knn.predict(X_test)

y_proba_knn = knn.predict_proba(X_test)[:, 1]


# ============================================================
# 16. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_pred_knn
)

print("\nConfusion Matrix:")
print(cm)


plt.figure(figsize=(5, 4))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['Class 0', 'Class 1'],
    yticklabels=['Class 0', 'Class 1']
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix - K-NN")

plt.show()


# ============================================================
# 17. CLASSIFICATION REPORT
# ============================================================

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred_knn
    )
)


# ============================================================
# 18. ACCURACY
# ============================================================

accuracy_knn = accuracy_score(
    y_test,
    y_pred_knn
)

print("\nAccuracy:")
print(accuracy_knn)

print("\nAccuracy percentage:")
print(f"{accuracy_knn * 100:.2f}%")


# ============================================================
# 19. PRECISION
# ============================================================

precision_knn = precision_score(
    y_test,
    y_pred_knn
)

print("\nPrecision:")
print(precision_knn)


# ============================================================
# 20. RECALL
# ============================================================

recall_knn = recall_score(
    y_test,
    y_pred_knn
)

print("\nRecall:")
print(recall_knn)


# ============================================================
# 21. ROC-AUC
# ============================================================

roc_auc_knn = roc_auc_score(
    y_test,
    y_proba_knn
)

print("\nROC-AUC:")
print(roc_auc_knn)


# ============================================================
# 22. ROC CURVE
# ============================================================

fpr_knn, tpr_knn, thresholds_knn = roc_curve(
    y_test,
    y_proba_knn
)

roc_auc = auc(
    fpr_knn,
    tpr_knn
)

plt.figure(figsize=(7, 5))

plt.plot(
    fpr_knn,
    tpr_knn,
    label=f'K-NN ROC curve (AUC = {roc_auc:.4f})'
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle='--'
)

plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')

plt.title('ROC Curve - Initial K-NN')

plt.legend(loc='lower right')

plt.show()


# ============================================================
# 23. GRID SEARCH FOR K-NN OPTIMIZATION
# ============================================================

param_grid = {
    'n_neighbors': range(1, 21),
    'weights': ['uniform', 'distance'],
    'metric': ['euclidean', 'manhattan']
}

grid_search = GridSearchCV(
    estimator=KNeighborsClassifier(),
    param_grid=param_grid,
    scoring='roc_auc',
    cv=5,
    n_jobs=-1
)

grid_search.fit(
    X_train,
    y_train
)


# ============================================================
# 24. BEST PARAMETERS
# ============================================================

print("\nBest parameters:")
print(grid_search.best_params_)

print("\nBest cross-validation ROC-AUC:")
print(grid_search.best_score_)


# ============================================================
# 25. GET BEST K-NN MODEL
# ============================================================

best_knn = grid_search.best_estimator_

print("\nBest K-NN model:")
print(best_knn)


# ============================================================
# 26. PREDICTIONS USING OPTIMIZED MODEL
# ============================================================

y_pred_best = best_knn.predict(
    X_test
)

y_proba_best = best_knn.predict_proba(
    X_test
)[:, 1]


# ============================================================
# 27. OPTIMIZED CONFUSION MATRIX
# ============================================================

cm_best = confusion_matrix(
    y_test,
    y_pred_best
)

print("\nOptimized Confusion Matrix:")
print(cm_best)


plt.figure(figsize=(5, 4))

sns.heatmap(
    cm_best,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['Class 0', 'Class 1'],
    yticklabels=['Class 0', 'Class 1']
)

plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.title(
    "Confusion Matrix - Optimized K-NN"
)

plt.show()


# ============================================================
# 28. OPTIMIZED CLASSIFICATION REPORT
# ============================================================

print("\nOptimized Classification Report:")

print(
    classification_report(
        y_test,
        y_pred_best
    )
)


# ============================================================
# 29. OPTIMIZED ACCURACY
# ============================================================

accuracy_best = accuracy_score(
    y_test,
    y_pred_best
)

print("\nOptimized Accuracy:")
print(accuracy_best)

print("\nOptimized Accuracy percentage:")
print(f"{accuracy_best * 100:.2f}%")


# ============================================================
# 30. OPTIMIZED PRECISION
# ============================================================

precision_best = precision_score(
    y_test,
    y_pred_best
)

print("\nOptimized Precision:")
print(precision_best)


# ============================================================
# 31. OPTIMIZED RECALL
# ============================================================

recall_best = recall_score(
    y_test,
    y_pred_best
)

print("\nOptimized Recall:")
print(recall_best)


# ============================================================
# 32. OPTIMIZED ROC-AUC
# ============================================================

roc_auc_best = roc_auc_score(
    y_test,
    y_proba_best
)

print("\nOptimized ROC-AUC:")
print(roc_auc_best)


# ============================================================
# 33. OPTIMIZED ROC CURVE
# ============================================================

fpr_best, tpr_best, thresholds_best = roc_curve(
    y_test,
    y_proba_best
)

roc_auc_best_curve = auc(
    fpr_best,
    tpr_best
)

plt.figure(figsize=(7, 5))

plt.plot(
    fpr_best,
    tpr_best,
    label=f'Optimized K-NN ROC curve (AUC = {roc_auc_best_curve:.4f})'
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle='--'
)

plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')

plt.title(
    'ROC Curve - Optimized K-NN'
)

plt.legend(loc='lower right')

plt.show()


# ============================================================
# 34. COMPARE INITIAL AND OPTIMIZED MODELS
# ============================================================

comparison = pd.DataFrame({
    'Model': [
        'Initial K-NN',
        'Optimized K-NN'
    ],

    'Accuracy': [
        accuracy_knn,
        accuracy_best
    ],

    'Precision': [
        precision_knn,
        precision_best
    ],

    'Recall': [
        recall_knn,
        recall_best
    ],

    'ROC-AUC': [
        roc_auc_knn,
        roc_auc_best
    ]
})

print("\nModel Comparison:")

display(comparison)


# ============================================================
# 35. FINAL RESULTS
# ============================================================

print("=" * 60)
print("FINAL RESULTS")
print("=" * 60)

print("\nInitial K-NN:")
print(f"Accuracy : {accuracy_knn:.4f}")
print(f"Precision: {precision_knn:.4f}")
print(f"Recall   : {recall_knn:.4f}")
print(f"ROC-AUC  : {roc_auc_knn:.4f}")

print("\nOptimized K-NN:")
print(f"Accuracy : {accuracy_best:.4f}")
print(f"Precision: {precision_best:.4f}")
print(f"Recall   : {recall_best:.4f}")
print(f"ROC-AUC  : {roc_auc_best:.4f}")

print("\nBest Parameters:")
print(grid_search.best_params_)

print("\nBest Cross-Validation ROC-AUC:")
print(grid_search.best_score_)

print("=" * 60)
```
