```python
# ============================================
# Water Potability Prediction Application
# ============================================

import streamlit as st
import pandas as pd
import joblib


# ============================================
# Load the saved model, scaler and features
# ============================================

model = joblib.load(
    "water_potability_model.pkl"
)

scaler = joblib.load(
    "water_potability_scaler.pkl"
)

features = joblib.load(
    "water_potability_features.pkl"
)


# ============================================
# Page Configuration
# ============================================

st.set_page_config(
    page_title="Water Potability Prediction",
    page_icon="💧",
    layout="centered"
)


# ============================================
# Application Title
# ============================================

st.title("💧 Water Potability Prediction")

st.write(
    "Enter the water-quality measurements below "
    "to predict whether the water is potable or not potable."
)


# ============================================
# Create input fields
# ============================================

ph = st.number_input(
    "pH",
    min_value=0.0,
    max_value=14.0,
    value=7.0
)

hardness = st.number_input(
    "Hardness",
    min_value=0.0,
    value=200.0
)

solids = st.number_input(
    "Solids",
    min_value=0.0,
    value=20000.0
)

chloramines = st.number_input(
    "Chloramines",
    min_value=0.0,
    value=7.0
)

sulfate = st.number_input(
    "Sulfate",
    min_value=0.0,
    value=300.0
)

conductivity = st.number_input(
    "Conductivity",
    min_value=0.0,
    value=400.0
)

organic_carbon = st.number_input(
    "Organic Carbon",
    min_value=0.0,
    value=10.0
)

trihalomethanes = st.number_input(
    "Trihalomethanes",
    min_value=0.0,
    value=60.0
)

turbidity = st.number_input(
    "Turbidity",
    min_value=0.0,
    value=4.0
)


# ============================================
# Prediction Button
# ============================================

if st.button("Predict Water Potability"):

    # Create a dictionary containing user input
    input_data = {
        "ph": ph,
        "Hardness": hardness,
        "Solids": solids,
        "Chloramines": chloramines,
        "Sulfate": sulfate,
        "Conductivity": conductivity,
        "Organic_carbon": organic_carbon,
        "Trihalomethanes": trihalomethanes,
        "Turbidity": turbidity
    }

    # Convert input into a DataFrame
    input_df = pd.DataFrame(
        [input_data]
    )

    # Make sure feature order matches training data
    input_df = input_df[features]

    # Scale the input
    input_scaled = scaler.transform(
        input_df
    )

    # Make prediction
    prediction = model.predict(
        input_scaled
    )

    # Get probability
    probability = model.predict_proba(
        input_scaled
    )[0][1]

    # ========================================
    # Display Prediction
    # ========================================

    st.subheader("Prediction Result")

    if prediction[0] == 1:

        st.success(
            "The water is predicted to be POTABLE."
        )

        st.write(
            f"Probability of Potable Water: "
            f"{probability * 100:.2f}%"
        )

    else:

        st.error(
            "The water is predicted to be NOT POTABLE."
        )

        st.write(
            f"Probability of Potable Water: "
            f"{probability * 100:.2f}%"
        )

    st.warning(
        "This is a machine-learning prediction and "
        "should not replace laboratory water-quality testing."
    )
```
