```python
# ============================================
# 1. IMPORT LIBRARIES
# ============================================

# Data manipulation and numerical operations
import pandas as pd
import numpy as np

# Data visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Data preprocessing and splitting
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer

# Machine Learning models
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier

# Model evaluation
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score,
    roc_curve,
    ConfusionMatrixDisplay
)

# Hyperparameter tuning
from sklearn.model_selection import GridSearchCV

# ============================================
# 2. LOAD THE DATASET
# ============================================

df = pd.read_excel("water_potability.xlsx")

print("Dataset loaded successfully!")
```

```python
# ============================================
# 3. INITIAL DATASET EXPLORATION
# ============================================

# Display the number of rows and columns
print("Dataset Shape:")
print(df.shape)

# Display the column names
print("\nDataset Columns:")
print(df.columns)
```

```python
# Display the first 5 rows of the dataset
print("\nFirst 5 Rows of the Dataset:")
print(df.head())
```

```python
# Display the last 5 rows of the dataset
print("\nLast 5 Rows of the Dataset:")
print(df.tail())
```

```python
# Display information about the dataset
print("\nDataset Information:")
df.info()
```

```python
# Display statistical summary of the dataset
print("\nStatistical Summary:")
print(df.describe())
```

```python
# 3.7 Check the distribution of the target variable
print("\nPotability Distribution:")
print(df["Potability"].value_counts())


# 3.8 Display target percentages
print("\nPotability Percentage:")
print(df["Potability"].value_counts(normalize=True) * 100)


# 3.9 Visualize the target distribution
plt.figure(figsize=(7, 5))

sns.countplot(x="Potability", data=df)

plt.title("Water Potability Distribution")
plt.xlabel("Potability (0 = Not Potable, 1 = Potable)")
plt.ylabel("Number of Samples")

plt.show()
```

```python
# ============================================
# 4. DATA QUALITY CHECK
# ============================================

# 4.1 Check for missing values
print("Missing Values in Each Column:")
print(df.isnull().sum())


# 4.2 Calculate the percentage of missing values
print("\nPercentage of Missing Values:")
missing_percentage = (df.isnull().sum() / len(df)) * 100
print(missing_percentage)


# 4.3 Check for duplicate rows
print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())


# 4.4 Check for unique values in the target column
print("\nUnique Potability Values:")
print(df["Potability"].unique())


# 4.5 Check for invalid negative values
numeric_columns = df.select_dtypes(include=np.number).columns

print("\nNegative Values in Each Numerical Column:")
print((df[numeric_columns] < 0).sum())


# 4.6 Display rows containing missing values
print("\nRows Containing Missing Values:")
print(df[df.isnull().any(axis=1)].head())
```

```python
# ============================================
# 5. DATA CLEANING AND PREPROCESSING
# ============================================

# 5.1 Create a copy of the original dataset
clean_df = df.copy()


# 5.2 Remove duplicate rows
duplicate_count = clean_df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

clean_df = clean_df.drop_duplicates()

print("Dataset shape after removing duplicates:")
print(clean_df.shape)


# 5.3 Check missing values before filling them
print("\nMissing values before imputation:")
print(clean_df.isnull().sum())


# 5.4 Separate features from the target
X = clean_df.drop("Potability", axis=1)
y = clean_df["Potability"]


# 5.5 Fill missing feature values using the median
imputer = SimpleImputer(strategy="median")

X_imputed = imputer.fit_transform(X)


# 5.6 Convert the processed features back into a DataFrame
X = pd.DataFrame(
    X_imputed,
    columns=X.columns,
    index=clean_df.index
)


# 5.7 Check that missing values have been handled
print("\nMissing values after imputation:")
print(X.isnull().sum())


# 5.8 Check the final target values
print("\nTarget values:")
print(y.value_counts())


# 5.9 Display the cleaned dataset
print("\nCleaned Dataset:")
print(X.head())
```

```python
# ============================================
# 6. EXPLORATORY DATA ANALYSIS (EDA)
# ============================================

# 6.1 Display the distribution of all numerical features

X.hist(
    figsize=(15, 12),
    bins=30,
    edgecolor="black"
)

plt.suptitle(
    "Distribution of Water Quality Features",
    fontsize=16
)

plt.tight_layout()
plt.show()


# 6.2 Box plots to identify possible outliers

plt.figure(figsize=(15, 8))

sns.boxplot(data=X)

plt.title("Box Plot of Water Quality Features")
plt.xlabel("Water Quality Features")
plt.ylabel("Values")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# 6.3 Correlation matrix

correlation_matrix = clean_df.corr(numeric_only=True)

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix of Water Quality Variables")

plt.tight_layout()
plt.show()


# 6.4 Correlation of each feature with Potability

potability_correlation = (
    correlation_matrix["Potability"]
    .sort_values(ascending=False)
)

print("\nCorrelation of Features with Potability:")
print(potability_correlation)


# 6.5 Visualize the relationship between features and Potability

features_to_plot = X.columns

for feature in features_to_plot:

    plt.figure(figsize=(7, 5))

    sns.boxplot(
        x=clean_df["Potability"],
        y=X[feature]
    )

    plt.title(f"{feature} vs Potability")
    plt.xlabel("Potability (0 = Not Potable, 1 = Potable)")
    plt.ylabel(feature)

    plt.tight_layout()
    plt.show()
```

```python
# ============================================
# 7. PREPARE FEATURES AND TARGET
# ============================================

# Define the input features
X = clean_df.drop("Potability", axis=1)

# Define the target variable
y = clean_df["Potability"]


# Display the feature names
print("Features used for Machine Learning:")
print(X.columns.tolist())


# Display the target variable
print("\nTarget Variable:")
print("Potability")


# Display the shape of the features and target
print("\nFeatures Shape:")
print(X.shape)

print("\nTarget Shape:")
print(y.shape)


# Display the first 5 feature records
print("\nFirst 5 Feature Records:")
print(X.head())


# Display the first 5 target values
print("\nFirst 5 Target Values:")
print(y.head())


# Display the number of samples in each target class
print("\nTarget Class Distribution:")
print(y.value_counts())
```

```python
# ============================================
# 8. TRAIN / TEST SPLIT
# ============================================

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


# Display the size of each dataset
print("Training Features Shape:", X_train.shape)
print("Testing Features Shape:", X_test.shape)

print("\nTraining Target Shape:", y_train.shape)
print("Testing Target Shape:", y_test.shape)


# Display the class distribution in the training set
print("\nTraining Target Distribution:")
print(y_train.value_counts())


# Display the class distribution in the testing set
print("\nTesting Target Distribution:")
print(y_test.value_counts())
```

```python
# ============================================
# 9. FEATURE SCALING
# ============================================

# Create the StandardScaler
scaler = StandardScaler()


# Fit the scaler using only the training data
# and transform the training data
X_train_scaled = scaler.fit_transform(X_train)


# Transform the testing data using the
# scaler learned from the training data
X_test_scaled = scaler.transform(X_test)


# Convert the scaled data back to DataFrames
X_train_scaled = pd.DataFrame(
    X_train_scaled,
    columns=X_train.columns,
    index=X_train.index
)

X_test_scaled = pd.DataFrame(
    X_test_scaled,
    columns=X_test.columns,
    index=X_test.index
)


# Display the first 5 rows of scaled training data
print("First 5 Rows of Scaled Training Data:")
print(X_train_scaled.head())


# Display basic statistics of the scaled data
print("\nStatistics of Scaled Training Data:")
print(X_train_scaled.describe())
```

```python
# ============================================
# 10. BUILD AND TRAIN THE LOGISTIC REGRESSION MODEL
# ============================================

# Create the Logistic Regression model
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)


# Train the model using the scaled training data
logistic_model.fit(
    X_train_scaled,
    y_train
)


# Display a message confirming that training is complete
print("Logistic Regression model trained successfully!")


# Display the model parameters
print("\nLogistic Regression Model:")
print(logistic_model)
```

```python
# ============================================
# 11. MAKE PREDICTIONS WITH LOGISTIC REGRESSION
# ============================================

# Make predictions on the testing data
y_pred_logistic = logistic_model.predict(X_test_scaled)


# Display the first 20 actual values
print("First 20 Actual Values:")
print(y_test.values[:20])


# Display the first 20 predicted values
print("\nFirst 20 Predicted Values:")
print(y_pred_logistic[:20])


# Compare actual and predicted values
comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred_logistic
})


print("\nActual vs Predicted:")
print(comparison.head(20))
```

```python
# ============================================
# 12. EVALUATE THE LOGISTIC REGRESSION MODEL
# ============================================

# Calculate evaluation metrics
accuracy_logistic = accuracy_score(y_test, y_pred_logistic)

precision_logistic = precision_score(
    y_test,
    y_pred_logistic,
    zero_division=0
)

recall_logistic = recall_score(
    y_test,
    y_pred_logistic,
    zero_division=0
)

f1_logistic = f1_score(
    y_test,
    y_pred_logistic,
    zero_division=0
)


# Get prediction probabilities for ROC-AUC
y_prob_logistic = logistic_model.predict_proba(
    X_test_scaled
)[:, 1]

roc_auc_logistic = roc_auc_score(
    y_test,
    y_prob_logistic
)


# Display the results
print("Logistic Regression Classification Results")
print("------------------------------------------")

print(f"Accuracy : {accuracy_logistic:.4f}")
print(f"Precision: {precision_logistic:.4f}")
print(f"Recall   : {recall_logistic:.4f}")
print(f"F1 Score : {f1_logistic:.4f}")
print(f"ROC-AUC  : {roc_auc_logistic:.4f}")


# Display the classification report
print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred_logistic,
        target_names=["Not Potable", "Potable"],
        zero_division=0
    )
)
```

```python
# ============================================
# 13. CONFUSION MATRIX
# ============================================

# Calculate the confusion matrix
cm_logistic = confusion_matrix(
    y_test,
    y_pred_logistic
)

# Display the numerical confusion matrix
print("Confusion Matrix:")
print(cm_logistic)


# Display the confusion matrix as a graph
plt.figure(figsize=(7, 5))

ConfusionMatrixDisplay(
    confusion_matrix=cm_logistic,
    display_labels=["Not Potable", "Potable"]
).plot()

plt.title("Confusion Matrix - Logistic Regression")

plt.tight_layout()
plt.show()
```

```python
# ============================================
# 14. ROC CURVE
# ============================================

# Calculate False Positive Rate and True Positive Rate
fpr_logistic, tpr_logistic, thresholds_logistic = roc_curve(
    y_test,
    y_prob_logistic
)


# Plot the ROC curve
plt.figure(figsize=(8, 6))

plt.plot(
    fpr_logistic,
    tpr_logistic,
    label=f"Logistic Regression (AUC = {roc_auc_logistic:.4f})"
)

# Plot the random classifier reference line
plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.title("ROC Curve - Logistic Regression")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# ============================================
# 15. BUILD AND TRAIN THE SVM MODEL
# ============================================

# Create the SVM model
svm_model = SVC(
    kernel="rbf",
    probability=True,
    random_state=42
)


# Train the SVM model using the scaled training data
svm_model.fit(
    X_train_scaled,
    y_train
)


# Display a message confirming that training is complete
print("SVM model trained successfully!")


# Display the model configuration
print("\nSVM Model:")
print(svm_model)
```

```python
# ============================================
# 16. MAKE PREDICTIONS WITH THE SVM MODEL
# ============================================

# Predict classes using the SVM model
y_pred_svm = svm_model.predict(X_test_scaled)


# Predict probabilities for ROC-AUC and ROC Curve
y_prob_svm = svm_model.predict_proba(X_test_scaled)[:, 1]


# Display the first 20 actual values
print("First 20 Actual Values:")
print(y_test.values[:20])


# Display the first 20 SVM predictions
print("\nFirst 20 SVM Predictions:")
print(y_pred_svm[:20])


# Create a comparison table
svm_comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred_svm
})


# Display comparison
print("\nActual vs SVM Predictions:")
print(svm_comparison.head(20))
```

```python
# ============================================
# 17. EVALUATE THE SVM MODEL
# ============================================

# Calculate evaluation metrics
accuracy_svm = accuracy_score(
    y_test,
    y_pred_svm
)

precision_svm = precision_score(
    y_test,
    y_pred_svm,
    zero_division=0
)

recall_svm = recall_score(
    y_test,
    y_pred_svm,
    zero_division=0
)

f1_svm = f1_score(
    y_test,
    y_pred_svm,
    zero_division=0
)


# Calculate ROC-AUC using SVM probabilities
roc_auc_svm = roc_auc_score(
    y_test,
    y_prob_svm
)


# Display the evaluation results
print("SVM Classification Results")
print("---------------------------")

print(f"Accuracy : {accuracy_svm:.4f}")
print(f"Precision: {precision_svm:.4f}")
print(f"Recall   : {recall_svm:.4f}")
print(f"F1 Score : {f1_svm:.4f}")
print(f"ROC-AUC  : {roc_auc_svm:.4f}")


# Display the classification report
print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred_svm,
        target_names=["Not Potable", "Potable"],
        zero_division=0
    )
)
```

```python
# ============================================
# 18. CONFUSION MATRIX - SVM
# ============================================

# Calculate the confusion matrix
cm_svm = confusion_matrix(
    y_test,
    y_pred_svm
)


# Display the numerical confusion matrix
print("SVM Confusion Matrix:")
print(cm_svm)


# Display the confusion matrix as a graph
plt.figure(figsize=(7, 5))

ConfusionMatrixDisplay(
    confusion_matrix=cm_svm,
    display_labels=["Not Potable", "Potable"]
).plot()

plt.title("Confusion Matrix - SVM")

plt.tight_layout()
plt.show()
```

```python
# ============================================
# 19. ROC CURVE - SVM
# ============================================

# Calculate False Positive Rate and True Positive Rate
fpr_svm, tpr_svm, thresholds_svm = roc_curve(
    y_test,
    y_prob_svm
)


# Plot the SVM ROC curve
plt.figure(figsize=(8, 6))

plt.plot(
    fpr_svm,
    tpr_svm,
    label=f"SVM (AUC = {roc_auc_svm:.4f})"
)


# Plot the random classifier reference line
plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)


# Add graph information
plt.title("ROC Curve - SVM")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
